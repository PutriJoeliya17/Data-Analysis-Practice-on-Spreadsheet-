
Program ini dibuat untuk melakukan automatisasi pengolahan (cleaning) data text yang berguna untuk pemodelan model analisa sentimen.


import pandas dan numpy untuk cleaning 

```python
import pandas as pd
import numpy as np
```

memasukan data csv ke vsc untuk di cleaning 

```python
df = pd.read_csv('osn.csv', low_memory=False)

df
```

melakukan penghapusan baris pada data yang kolom kelasnya kosong atau missing value 

```python
df = df.dropna(subset=['Kelas'])
```

melakukan pengisian data pada baring baring missing value pada kolom prize tambahan 

```python
df['Prize Tambahan'] = df['Prize Tambahan'].fillna('Tidak Ada')
```

melihat jumlah data, keterangan tipe data agar sesuai 

```python
df.info()
```

menghilangkan duplikat pada data (jika ada)

```python
df = df.drop_duplicates()
```

melihat panjangnya data 

```python
print(len(df))
```

data final setelah di cleaning 

```python
df
```

mengubah data yang sudah clean kedalam csv untuk selanjutnya di kerjakan seluruhnya dalam spreadsheet 

```python
df.to_csv('osn_data_cleaned.csv', index=False)
```
